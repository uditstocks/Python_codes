### 14. Generate the First n Prime Numbers Using a Generator

```python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, num):
        if num%i == 0:
            return False
    return True

def prime_generator(n):
    num = 2
    while n:
        if is_prime(num):
            yield num
            n-=1
        num+=1
        
```

```python
x = int(input("enter how many prime numbers you want: "))
result = prime_generator(x)
for e in result:
    print(e, end = " ")
```

```python
num1 = int(input("Enter 1st number: "))
num2 = int(input("Enter 2nd number: "))

def gcd(a,b):
    if a == 0:
        return b
    if b == 0:
        return a
    if a == b:
        return a
    elif a > b:
        return gcd(a - b, b)
    else:
        return gcd(a, b -a)

print(gcd(num1,num2))
```

### 30. Sandglass Pattern

```python
def sand(n):
    for i in range(n):
        # spaces (increase)
        for j in range(i):
            print(" ",end="")
        # stars (decrease)
        for j in range(n-i):
            print("*",end=" ")
        print()

    for i in range(n):
        for j in range(i,n-1):
            print(" ", end="")
        for j in range(i+1):
            print("*", end=" ")
        print()
```

```python
sand(5)
```

### 49. Convert a Roman Numeral to an Integer

Convert a Roman numeral string into its equivalent integer value, following the addition and subtraction rules
(e.g. `IV` = 4, `XI` = 11).

```python
class Solution(object):
    def romanToInt(self, s):
        values = {
            'I': 1, 'V': 5, 'X': 10,
            'L': 50, 'C': 100,
            'D': 500, 'M': 1000
        }

        total = 0

        for i in range(len(s)):
            if i < len(s)-1 and values[s[i]] < values[s[i+1]]:
                total -= values[s[i]]
            else:
                total += values[s[i]]

        return total
```

```python
a = Solution()
a.romanToInt("XI")
```

```python
def secondlargest(arr): 
    largest = float('-inf')   
    secondlargest = float('-inf')  
 
    for num in arr: 
        if num > largest: 
            secondlargest = largest
            largest = num  

        if num < largest and num > secondlargest:
            secondlargest = num

    return secondlargest
```

### 51. Binary Search (Iterative)

**Key points:**
- Works only on a **sorted** array.
- **Divide & conquer:** check the middle element and eliminate half of the search space at each step.
- **Pointer movement:**
  - If target < middle → search the left half.
  - If target > middle → search the right half.

```python
def binarysearch(arr, element):
    start = 0
    end = len(arr) - 1
    while(start <= end):
        mid = (start + end) // 2
        if arr[mid] == element:
            return mid
        elif arr[mid] < element:
            start = mid + 1
        else:
            end = mid - 1
```

```python
arr = [1,2,3,4,5]
index = binarysearch(arr, 5)
print(index)
```

### 52. Binary Search (Recursive)

```python
def binarySearch(a, x, si, ei): #here "a" is the list and "x" is the item to find
    if si > ei:
        return -1

    mid = (si + ei)//2
    if a[mid] == x:
        return mid
    elif a[mid] > x:
        return binarySearch(a, x, si, mid - 1)
    else:
        return binarySearch(a, x, mid + 1, ei)

y = binarySearch([1,2,3,4,5,6,7,8,9], 3, 0, 8)
print(y)
```

### 53. Selection Sort

**Key points:**
- For each position, find the smallest element in the rest of the array and place it there.
- Works on an unsorted array.
- **Goal:** put the correct element at position `i` in every step.
- **Strategy:** find the minimum of the remaining array → swap it to the front.

```python
def selectionshort(arr):
    length = len(arr)
    #put the correct element at "i"th position 
    for i in range(length - 1):
        minIndex = i
        #calculating the index of minimum element
        for j in range(i + 1, length):
            if(arr[j] < arr[minIndex]):
                minIndex = j
        arr[i],arr[minIndex] = arr[minIndex],arr[i]
    
arr = [1,3,6,2,4,9,8,5,7,45]
selectionshort(arr)
print(arr)
```

### 54. Bubble Sort

**Key points:**
- Works on any array.
- Repeatedly compares adjacent elements.
- If left > right → swap them.
- After each pass, the largest remaining element moves to the end.

```python
def bubbleSort(arr):
    length = len(arr)
    # i runs for n - 1 rounds
    # j: in each round, go up to position length - 1 - i
    for i in range(length - 1):
        for j in range(length - 1 - i):
            if (arr[j] > arr[j + 1]):
                arr[j],arr[j + 1] = arr[j + 1], arr[j]
                
        

arr = [1,11,8,5,20,2]
bubbleSort(arr)
print(arr)
```

### 55. Insertion Sort

**Key points:**
- Take one element → insert it at its correct position.
- Keep shifting bigger elements to the right until the correct place is found.
- Builds the sorted portion step by step.
- Works on any array.

```python
def insertionSort(arr):
    length = len(arr)
    for i in range(1, length):
        j = i - 1
        temp = arr[i]
        # Shift elements while this condition holds
        while(j>=0 and arr[j] > temp):
            arr[j + 1] = arr[j]
            j = j - 1
        # j + 1 is the correct position for the ith element
        arr[j + 1] = temp

arr = [1,11,8,5,20,2,12]
insertionSort(arr)
print(arr)
```

### 56. Merge Two Sorted Arrays

**Key points:**
- **Two pointers:** always pick the smaller front element.
- **No sorting needed:** the arrays are already sorted, just merge them.
- **Linear time:** O(n + m).

```python
def mergeSortedArrays(arr1, arr2):
    i = 0
    j = 0
    len1 = len(arr1)
    len2 = len(arr2)
    arr = []

    while ((i < len1) and (j < len2)):
        if(arr1[i] < arr2[j]):
            arr.append(arr1[i])
            i = i + 1
        else:
            arr.append(arr2[j])
            j = j + 1
    while (i < len1):
        arr.append(arr1[i])
        i = i + 1
    while (j < len2):
        arr.append(arr2[j])

        j = j + 1
    return arr
        
arr1 =[1, 3, 5, 8, 12, 19, 21]
arr2 =[2, 4, 6, 9, 11, 17, 22]
arr = mergeSortedArrays(arr1,arr2)
print(arr)
```

### 57. Remove Consecutive Duplicates from a String (Recursion)

Example: `"aabbsderrrwwqq"` → `"absderwq"`

```python
def removeConsecutiveDuplicates(s):
    if len(s) == 1:
        return s

    if s[0] == s[1]:
        return removeConsecutiveDuplicates(s[1:])
    else:
        return s[0] + removeConsecutiveDuplicates(s[1:])
```

```python
y = removeConsecutiveDuplicates("aabbsderrrwwqq")
print(y)
```

### 58. Longest Substring Without Repeating Characters

Given a string `s`, return the length of the longest substring that contains no repeating characters.

**Examples:**

| Input | Output | Explanation |
|---|:---:|---|
| `"abcabcbb"` | 3 | The substring `"abc"` |
| `"bbbbb"` | 1 | The substring `"b"` |
| `"pwwkew"` | 3 | The substring `"wke"` |
| `""` | 0 | Empty string |

```python
def length_of_longest_substring(s: str) -> int:
    last_seen = {}   # char -> most recent index it appeared at
    start = 0        # left edge of the current window
    best = 0

    for i, ch in enumerate(s):
        # Only jump 'start' forward if ch repeats *inside* the window.
        if ch in last_seen and last_seen[ch] >= start:
            start = last_seen[ch] + 1
        last_seen[ch] = i
        best = max(best, i - start + 1)

    return best
```

#### How It Works

Keep a window `[start, i]` that always holds distinct characters. As you scan right with `i`, record each
character's most recent index in `last_seen`. When you hit a character you've seen before, slide `start` to just
past its previous occurrence, which drops the duplicate out of the window. The answer is the largest window width
seen along the way.

The condition `last_seen[ch] >= start` is the part people often miss. A character can exist in `last_seen` but sit
to the left of the current window, so it isn't actually a duplicate anymore and `start` shouldn't move backward.
Try `"abba"`: when you reach the final `'a'`, its old index `0` is behind `start` (which is `2` by then), so you
correctly ignore it and get `2` rather than a buggy smaller value.

#### Complexity

- **Time:** O(n), since each character is visited once.
- **Space:** O(min(n, k)), where k is the alphabet size, for the `last_seen` map.

   In Python, prefix attributes with `_` (protected) or `__` (private) to restrict access.
2. **Inheritance:** A child class reuses and extends a parent class.
   *Like a child inheriting traits from a parent.*
   The child gets all of the parent's methods and can add its own or override existing ones.
3. **Polymorphism:** Same method name, different behavior per class.
   *Like a "speak" command: a dog barks, a cat meows, a cow moos.*
   Same interface, different implementation. Achieved via method overriding in Python.
4. **Abstraction:** Hiding complexity; showing only essential details.
   *Like driving a car: you use the steering wheel and pedals without knowing how the engine works.*
   In Python, done via abstract base classes (`abc` module).

#### Inheritance: A Child Class Extends a Parent Class

The child class gets all methods and attributes of the parent. Use `super()` to call the parent's `__init__`
without repeating code.

```python
# Single inheritance
class Animal:
    def __init__(self, name, sound):
        self.name  = name
        self.sound = sound

    def speak(self):
        return f"{self.name} says {self.sound}!"

    def breathe(self):
        return f"{self.name} breathes air."

class Dog(Animal):              # inherits from Animal
    def __init__(self, name):
        super().__init__(name, "Woof")   # call parent __init__

    def fetch(self):            # new method only in Dog
        return f"{self.name} fetches the ball!"

d = Dog("Bruno")
print(d.speak())                # inherited from Animal
print(d.breathe())              # inherited from Animal
print(d.fetch())                # Dog-specific method
print(isinstance(d, Animal))    # True - Dog IS-A Animal
```

```python
# Multi-level inheritance
class Vehicle:
    def start(self):
        return "Vehicle started"

class Car(Vehicle):
    def drive(self):
        return "Car is driving"

class ElectricCar(Car):         # inherits from Car → Vehicle
    def charge(self):
        return "Charging battery"

ec = ElectricCar()
print(ec.start())               # from Vehicle (grandparent)
print(ec.drive())               # from Car (parent)
print(ec.charge())              # own method
print(ElectricCar.__mro__)      # method resolution order
```

```python
# Multiple inheritance
class Flyable:
    def fly(self):
        return "I can fly!"

class Swimmable:
    def swim(self):
        return "I can swim!"

class Duck(Flyable, Swimmable):  # inherits from TWO classes
    def quack(self):
        return "Quack!"

d = Duck()
print(d.fly())
print(d.swim())
print(d.quack())
```

#### Encapsulation: Controlling Access to Data

Python uses naming conventions to signal access levels. Unlike Java, it doesn't enforce them at the language
level, but the convention is respected.

| Access level | Syntax | Meaning |
|---|---|---|
| Public | `self.name` | Accessible everywhere |
| Protected | `self._salary` | Convention: don't access outside the class/subclass |
| Private | `self.__pin` | Name-mangled; hard to access from outside |

##### Example: `BankAccount`

The last line raises `AttributeError` on purpose, because private attributes are name-mangled and can't be accessed directly.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner    = owner        # public
        self._limit   = 50000        # protected
        self.__balance = balance     # private (name-mangled)

    def deposit(self, amt):
        if amt > 0:
            self.__balance += amt

    def withdraw(self, amt):
        if amt > self.__balance:
            return "Insufficient funds"
        self.__balance -= amt
        return f"Withdrawn ₹{amt}"

    def get_balance(self):           # getter - controlled access
        return self.__balance

acc = BankAccount("Rahul", 10000)
acc.deposit(5000)
print(acc.get_balance())             # 15000 - via getter
print(acc.withdraw(3000))            # Withdrawn ₹3000
print(acc.owner)                     # public - ok
print(acc._limit)                    # works but discouraged
print(acc.__balance)                 # AttributeError!
```

##### Example: Pythonic Getters & Setters with `@property`

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary

    @property
    def salary(self):               # getter
        return self.__salary

    @salary.setter
    def salary(self, value):        # setter with validation
        if value < 0:
            raise ValueError("Salary can't be negative")
        self.__salary = value

e = Employee("Priya", 50000)
print(e.salary)                     # 50000 - getter
e.salary = 60000                    # setter
print(e.salary)                     # 60000
```

#### Polymorphism: One Interface, Many Forms

The same method name behaves differently depending on the object calling it. This makes code flexible and
extensible: you can add new classes without changing existing logic.

##### Method Overriding

```python
class Shape:
    def area(self):
        return 0

    def describe(self):
        return f"I am a {type(self).__name__} with area {self.area()}"

class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):              # overrides parent method
        return round(3.14 * self.r ** 2, 2)

class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w; self.h = h
    def area(self):              # overrides parent method
        return self.w * self.h

class Triangle(Shape):
    def __init__(self, b, h):
        self.b = b; self.h = h
    def area(self):
        return 0.5 * self.b * self.h

shapes = [Circle(5), Rectangle(4, 6), Triangle(3, 8)]
for s in shapes:
    print(s.describe())          # same method, different result
```

#### Dunder (Special) Methods

Methods wrapped in double underscores (`__method__`) are called "dunder" methods. Python calls them automatically,
letting your objects work with built-in operations like `print()`, `len()`, `+`, `==` and more.

```python
# A Python class demonstrating operator overloading and built-in function customization using dunder methods.
class Book:
    def __init__(self, title, author, pages):
        self.title  = title
        self.author = author
        self.pages  = pages

    def __str__(self):           # called by print() and str()
        return f'"{self.title}" by {self.author}'

    def __repr__(self):          # called in REPL, debugging
        return f'Book("{self.title}", "{self.author}", {self.pages})'

    def __len__(self):           # called by len()
        return self.pages

    def __eq__(self, other):     # called by ==
        return self.title == other.title

    def __lt__(self, other):     # called by <  (enables sorting)
        return self.pages < other.pages

    def __add__(self, other):    # called by +
        combined = self.pages + other.pages
        return f"Combined pages: {combined}"

b1 = Book("Python 101", "Rahul", 350)
b2 = Book("OOP Guide", "Priya", 200)

print(b1)                        # uses __str__
print(len(b1))                   # uses __len__
print(b1 == b2)                  # uses __eq__
print(b2 < b1)                   # uses __lt__
print(b1 + b2)                   # uses __add__
books = [b1, b2]
print(sorted(books))             # sorts using __lt__
```

#### Instance Methods vs `@classmethod` vs `@staticmethod`

```python
# A Python class demonstrating the difference between instance methods, class methods, and static methods.
class Temperature:
    unit = "Celsius"

    def __init__(self, value):
        self.value = value

    @classmethod
    def set_unit(cls, unit):         # accesses the class, not instance
        cls.unit = unit

    @staticmethod
    def celsius_to_fahrenheit(c):    # utility - no class or instance needed
        return (c * 9/5) + 32

    def display(self):
        return f"{self.value}° {Temperature.unit}"

t = Temperature(100)
print(t.display())                   # 100° Celsius
Temperature.set_unit("Kelvin")
print(t.display())                   # 100° Kelvin
print(Temperature.celsius_to_fahrenheit(100))  # 212.0
```

### 60. Abstract Classes & Abstraction (ABC)

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

    def describe(self):  # concrete method - shared by all subclasses
        print(f"I am a shape with area = {self.area():.2f}")


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

    def perimeter(self):
        return 2 * 3.14159 * self.radius


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


c = Circle(5)
r = Rectangle(4, 6)

print(f"Circle    -> Area: {c.area():.2f}, Perimeter: {c.perimeter():.2f}")
c.describe()

print(f"Rectangle -> Area: {r.area()}, Perimeter: {r.perimeter()}")

# Uncommenting below raises TypeError - abstract class can't be instantiated
# s = Shape()
```

#### What's Happening Here?

- `Shape` is an **abstract class**, so it cannot be instantiated directly.
- `@abstractmethod` forces every subclass to implement `area()` and `perimeter()`. If they don't, Python raises a `TypeError`.
- An ABC can also have **concrete methods** (like `describe()`), unlike a pure interface.
- This enforces a **contract**: any class claiming to be a `Shape` must define how it calculates area and perimeter.

> **Interview angle:** "Abstraction hides *how* something works and exposes *what* it does. ABC enforces that contract at the class level."

### 61. MRO: Method Resolution Order

```python
class A:
    def greet(self):
        print("Hello from A")

class B(A):
    def greet(self):
        print("Hello from B")

class C(A):
    def greet(self):
        print("Hello from C")

class D(B, C):  # Diamond: D -> B -> C -> A -> object
    pass


d = D()
d.greet()   # Which greet() gets called?

print(f"MRO: {[cls.__name__ for cls in D.__mro__]}")


# Bonus: super() walks the MRO chain
class E(B, C):
    def greet(self):
        super().greet()   # calls B.greet() → next in MRO
        print("Hello from E")

print("---")
e = E()
e.greet()
```

#### What's Happening Here?

- Python resolves method lookup using **C3 linearization**: left to right, depth-first, no class visited twice.
- `D(B, C)` → MRO is `[D, B, C, A, object]`, so `d.greet()` calls `B.greet()` (B comes before C).
- `super()` doesn't just call the parent; it calls the **next class in the MRO chain**, which is why it's safe in multiple inheritance.
- Use `ClassName.__mro__` to debug which method will get called.

> **Interview angle:** "The diamond problem in Python is solved by MRO. Python guarantees each class appears only once in the chain, and `super()` respects that order."

### 62. Decorators

A decorator wraps a function to add behavior without changing its code.

Think of it like a gift wrapper: the gift (function) stays the same inside, but the wrapper adds something extra
on the outside. In Python, a decorator is itself a function that takes a function as input and returns a new
(wrapped) function.

#### Core Examples

##### 1. Timer: Measure Execution Time

```python
# measure execution time
import time
from functools import wraps

def timer(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} took {(end-start):.4f} seconds")
        return result
    return wrapper

@timer
def slow_task():
    time.sleep(1)
    return "Done"

slow_task()
```

##### 2. Validator: Check Inputs Before Running

The second call raises `ValueError` on purpose to show the validation in action.

```python
# check inputs before running
def validate_positive(func):
    @wraps(func)
    def wrapper(n):
        if n < 0:
            raise ValueError(f"Expected positive number, got {n}")
        return func(n)
    return wrapper

@validate_positive
def square_root(n):
    return n ** 0.5

print(square_root(16))    # 4.0
print(square_root(-4))    # raises ValueError
```

##### 3. Logger: Track When Functions Are Called

```python
from functools import wraps

def logger(func):
    @wraps(func)               # preserves original function name & docstring
    def wrapper(*args, **kwargs):
        print(f"Calling: {func.__name__} | args={args}")
        result = func(*args, **kwargs)
        print(f"Returned: {result}")
        return result
    return wrapper

@logger
def add(a, b):
    return a + b

add(3, 5)
```

#### Real-World Use Cases

##### 1. Authentication Check (used in Flask/Django)

```python
def login_required(func):
    @wraps(func)
    def wrapper(user, *args, **kwargs):
        if not user.get("logged_in"):
            print("Access denied! Please log in.")
            return None
        return func(user, *args, **kwargs)
    return wrapper

@login_required
def dashboard(user):
    print(f"Welcome, {user['name']}!")

user1 = {"name": "Rahul", "logged_in": True}
user2 = {"name": "Guest", "logged_in": False}

dashboard(user1)    # Welcome, Rahul!
dashboard(user2)    # Access denied!
```

##### 2. Retry on Failure (Network/API Calls)

```python
import time
from functools import wraps

def retry(times=3, delay=1):         # decorator with arguments
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            for attempt in range(1, times+1):
                try:
                    return func(*args, **kwargs)
                except Exception as e:
                    print(f"Attempt {attempt} failed: {e}")
                    time.sleep(delay)
            print("All attempts failed.")
        return wrapper
    return decorator

@retry(times=3, delay=0.5)
def fetch_data():
    raise ConnectionError("Server down")

fetch_data()
```

##### 3. Cache / Memoization (Speed Up Repeated Calls)

```python
from functools import lru_cache

@lru_cache(maxsize=128)     # built-in caching decorator
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(10))   # 55
print(fibonacci(50))   # 12586269025 - instant thanks to caching
```

#### Chaining Decorators

When you stack decorators, they are applied bottom-up: the one closest to the function is applied first.
The order matters.

```python
from functools import wraps

def bold(func):
    @wraps(func)
    def wrapper():
        return "<b>" + func() + "</b>"
    return wrapper

def italic(func):
    @wraps(func)
    def wrapper():
        return "<i>" + func() + "</i>"
    return wrapper

@bold          # applied second (outer)
@italic        # applied first (inner)
def greet():
    return "Hello"

print(greet())
# Equivalent to: bold(italic(greet))()
```

#### Class-Based Decorators

Decorators can also be implemented as classes using `__init__` and `__call__`.

```python
class CountCalls:
    def __init__(self, func):
        self.func = func
        self.count = 0

    def __call__(self, *args, **kwargs):
        self.count += 1
        print(f"Call #{self.count} to {self.func.__name__}")
        return self.func(*args, **kwargs)

@CountCalls
def say_hi():
    print("Hi!")

say_hi()
say_hi()
say_hi()
```

#### Common Interview Questions on Decorators

These are the most commonly asked decorator questions in Python interviews at all levels.

**Q1. What is a decorator in Python?**

**Answer:** A decorator is a design pattern in Python that allows you to add functionality to an existing function
without modifying its code. It is a function that takes another function as an argument, wraps it inside an inner
function, and returns the wrapper. The `@` syntax is syntactic sugar for reassigning the function to its decorated
version.

**Q2. Why do we use `@wraps` when writing a decorator?**

**Answer:** When you wrap a function, Python replaces the original function object with the wrapper. This means the
original function's `__name__`, `__doc__` and other metadata are lost. Using `@wraps(func)` from the `functools`
module copies those attributes from the original function to the wrapper, which is important for debugging and
documentation.

```python
def dec(func):
    @wraps(func)              # without this: wrapper.__name__ = "wrapper"
    def wrapper(*args, **kwargs):   # with this: wrapper.__name__ = "add"
        return func(*args, **kwargs)
    return wrapper
```

**Q3. Can a function have multiple decorators? What is the execution order?**

**Answer:** Yes. Multiple decorators are applied bottom-up (innermost first), but they execute top-down at call time.
So `@bold` `@italic` on a function means: `italic` is applied first during decoration, but `bold`'s wrapper runs
first when the function is called.

**Q4. What is the difference between `@staticmethod` and `@classmethod`?**

**Answer:**
- `@staticmethod`: no access to the class (`cls`) or the instance (`self`). It's just a regular function scoped to the class namespace.
- `@classmethod`: receives the class (`cls`) as the first argument instead of `self`. Used to create factory methods or access class-level data.

```python
class MyClass:
    count = 0

    @classmethod
    def get_count(cls):        # accesses class
        return cls.count

    @staticmethod
    def utility(x):            # no class/instance access
        return x * 2
```

**Q5. What is the `@property` decorator?**

**Answer:** `@property` turns a method into a readable attribute. It allows you to add validation or computed logic
behind attribute-style access. Pair it with `@name.setter` to allow setting the value.

```python
class Circle:
    def __init__(self, radius):
        self._radius = radius

    @property
    def area(self):
        return 3.14 * self._radius ** 2

c = Circle(5)
print(c.area)   # 78.5 - called like attribute, not method
```

### 63. Path Sum

Given the `root` of a binary tree and an integer `targetSum`, return `True` if the tree has a **root-to-leaf** path
such that adding up all the values along the path equals `targetSum`.

```python
class Solution:
    def hasPathSum(self, root: Optional[TreeNode], targetSum: int) -> bool:
        # node hi nahi hai, to raasta bhi nahi
        if not root:
            return False

        # is node ka paisa kaato (remaining = bache hue paise)
        remaining = targetSum - root.val

        # dead end (leaf) aa gaya, to check karo: jeb khaali hai kya?
        if not root.left and not root.right:
            return remaining == 0

        # dead end nahi hai, to bache hue paise leke aage jao:
        # pehle left mein try karo, wahan nahi mila to right mein
        return (self.hasPathSum(root.left, remaining) or
                self.hasPathSum(root.right, remaining))
```

### 64. Same Tree

Given the roots of two binary trees `p` and `q`, write a function to check whether they are the same.

Two binary trees are considered the same if they are structurally identical and the nodes have the same values.

```python
# Definition for a binary tree node.
class TreeNode(object):
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution(object):
    def isSameTree(self, p ,q):

        # Both trees are empty
        if p is None and q is None:
            return True

        # one is empty and other isn't
        if p is None or q is None:
            return False

        # values don't match
        if p.val != q.val:
            return False

        return (

            self.isSameTree(p.left, q.left)
            and
            self.isSameTree(p.right, q.right)
        )
        
```

### 65. Symmetric Tree

Given the `root` of a binary tree, check whether it is a mirror of itself (i.e., symmetric around its center).

```python
# Definition for a binary tree node.
class TreeNode(object):
     def __init__(self, val=0, left=None, right=None):
         self.val = val
         self.left = left
         self.right = right
class Solution(object):
    def isSymmetric(self, root):
        def mirror(left, right):
            if left is None and right is None:
                return True
            if left is None or right is None:
                return False
            if left.val != right.val:
                return False
            
            return(
                mirror(left.left, right.right)
                and
                mirror(left.right, right.left)
            )
            
        return mirror(root.left, root.right)    
```

### 66. Height of a Binary Tree

Find the height of a binary tree, i.e. the number of nodes on the longest path from the root down to a leaf.

```python
def height(root):
    if root is None:
        return 0
    heightleft = height(root.left)
    heightright = height(root.right)
    return 1 + max(heightleft, heightright)

    
```

### 67. Count Nodes Greater Than X

Given the `root` of a binary tree and an integer `X`, count the nodes whose value is greater than `X`.

```python
def countNodesGreaterThanX(root, X):
    current = 0
    if root is None:
        return 0
    if root.val > X:
        current = 1
    left = countNodesGreaterThanX(root.left, X)
    right = countNodesGreaterThanX(root.right, X)
    return current + left + right
    
```

### 68. Count Leaf Nodes

Given the `root` of a binary tree, count and return the total number of leaf nodes.
A leaf node is a node that has no left child and no right child.

```python
def countLeafNodes(root):
    # Empty tree has no leaf nodes
    if root is None:
        return 0

    # Current node is a leaf
    if root.left is None and root.right is None:
        return 1

    # Count leaves in both subtrees
    left = countLeafNodes(root.left)
    right = countLeafNodes(root.right)

    # Combine results
    return left + right
```
